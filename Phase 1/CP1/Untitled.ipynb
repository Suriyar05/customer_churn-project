```python
df["TotalCharges"]=pd.to_numeric(df["TotalCharges"],errors="coerce")
```

```python
#4
columns = ["gender", "Partner", "Dependents",
         "PhoneService", "InternetService", "Contract", "PaymentMethod", "Churn"]
 
for col in columns:
    print(f"========== {col} ==========")
    print("Unique Values :", ", ".join(df[col].unique().tolist()))
    print()
    print("Value Count")
    for value, count in df[col].value_counts().items():
        print(f"{value}: {count}")
    print()
    
```
